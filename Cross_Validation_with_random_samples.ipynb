```python
import numpy as np
import xgboost as xgb
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split, KFold, cross_val_score, StratifiedKFold
import warnings
warnings.filterwarnings('ignore')
```

```python
#Generating synthetic data & splitting into train-test
x , y = make_classification(n_samples=1000,n_informative=2,n_features=20, random_state=42)
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.3,random_state=42)
```

```python
# XGBoost Classifier

clf = xgb.XGBClassifier(use_label_encoder=False, eval_metric='logloss')
clf.fit(x_train, y_train)
```

```python
#Perform k-Fold Cross-Validation
kf = KFold (n_splits=5, shuffle=True, random_state=42)
kfold_scores = cross_val_score(clf, x, y, cv=kf, )
```

```python
kfold_scores
```

```python
#Perform Stratified k-Fold Cross-Validation
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
skfold_scores = cross_val_score(clf, x, y, cv=skf)
```
